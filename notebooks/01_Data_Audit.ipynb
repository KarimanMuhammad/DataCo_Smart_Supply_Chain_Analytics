```python
date_columns = [
    "order date (DateOrders)",
    "shipping date (DateOrders)"
]

for col in date_columns:
    df[col] = pd.to_datetime(df[col], errors="coerce")
```

## 7. Dataset Grain

Each row represents an individual order item rather than a complete order.

Therefore, order-level KPIs must be calculated using distinct Order Id values,
while item-level metrics can be calculated using the individual records.

```python
grain_summary = {
    "Total Rows": len(df),
    "Unique Orders": df["Order Id"].nunique(),
    "Unique Order Items": df["Order Item Id"].nunique(),
    "Unique Customers": df["Customer Id"].nunique(),
    "Unique Products": df["Product Card Id"].nunique()
}

pd.Series(grain_summary)
```

## 8. Business Data Profile

This section summarizes the main business dimensions of the dataset,
including orders, customers, products, sales, profit, and quantity.

```python
business_profile = {
    "Total Sales": df["Sales"].sum(),
    "Total Profit": df["Benefit per order"].sum(),
    "Total Orders": df["Order Id"].nunique(),
    "Total Customers": df["Customer Id"].nunique(),
    "Total Products": df["Product Card Id"].nunique(),
    "Total Quantity": df["Order Item Quantity"].sum()
}

pd.Series(business_profile)
```
