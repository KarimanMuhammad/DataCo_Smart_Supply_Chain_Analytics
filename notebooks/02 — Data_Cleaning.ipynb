# Data Cleaning & Preparation

This section prepares the raw dataset for analysis by removing irrelevant
fields, handling missing values, correcting data types, and creating a clean
dataset for downstream analysis.

## 1. Load Raw Dataset

```python
import pandas as pd
import numpy as np

```

```python
file_path = r"D:\DataCo_Smart_Supply_Chain_Analytics\data\raw\DataCoSupplyChainDataset.csv"

df = pd.read_csv(file_path)
```

## 2. Date Conversion

```python
date_columns = [
    "order date (DateOrders)",
    "shipping date (DateOrders)"
]

for col in date_columns:
    df[col] = pd.to_datetime(df[col], errors="coerce")
```

## 3. Remove Irrelevant and Sensitive Columns

The following fields are excluded because they are either sensitive,
not required for business analysis, or contain no analytical value.

```python
columns_to_drop = [
    "Customer Password",
    "Customer Email",
    "Customer Fname",
    "Customer Lname",
    "Customer Street",
    "Product Description",
    "Product Image"
]

df = df.drop(columns=columns_to_drop)
```

## 4. Missing Value Assessment

Missing values are reviewed after removing irrelevant fields to determine
which remaining variables require treatment.

```python
missing = (
    df.isna()
      .sum()
      .sort_values(ascending=False)
)

missing[missing > 0]
```

## 5. Handle Missing Values

Order Zipcode contains a very high proportion of missing values and is not
required for the geographic analysis because location can be analyzed using
country, state, city, and region fields.

Customer Zipcode contains only a few missing values but is not required for
the business analysis.

Both columns are therefore excluded rather than artificially imputing values.

```python
columns_to_drop = [
    "Order Zipcode",
    "Customer Zipcode"
]

df = df.drop(columns=columns_to_drop)
```

```python
df.isna().sum().sort_values(ascending=False).head()
```

## 6. Final Data Quality Check

The cleaned dataset is checked to confirm that irrelevant fields and
unresolved missing values have been removed before feature engineering.

```python
print("Shape:", df.shape)
print("Missing values:", df.isna().sum().sum())
print("Duplicate rows:", df.duplicated().sum())
```

```python
output_path = r"D:\DataCo_Smart_Supply_Chain_Analytics\data\processed\DataCo_cleaned.csv"

df.to_csv(output_path, index=False)
```
